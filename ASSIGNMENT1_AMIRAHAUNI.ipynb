```python
# Python String capitalize() Method
# The capitalize() method returns a string where the first character is upper case, and the rest is lower case.

# https://www.w3schools.com/python/ref_string_capitalize.asp

# EXAMPLE 1
txt = "hello, and welcome to my world."
x = txt.capitalize()
print (x)

# EXAMPLE 2
meals = "my Father's tomyum is the best!"
m = meals.capitalize()
print(m)

# EXAMPLE 3
shoe = "makerz is one of the Best LOCAL shoes brand."
s = shoe.capitalize()
print(s)
```

```python
# Python String casefold() Method
# The casefold() method returns a string where all the characters are lower case.

# https://www.w3schools.com/python/ref_string_casefold.asp

# EXAMPLE 1
txt = "Hello, And Welcome To My World!"
x = txt.casefold()
print(x)

# EXAMPLE 2
run = "I run 3KM everyday."
r = run.casefold()
print(r)

# EXAMPLE 3
car = "The Toyota YARIS model attracts Many Youngsters."
c = car.casefold()
print(c)
```

```python
# Python String count() Method
# The count() method returns the number of times a specified value appears in the string.

# https://www.w3schools.com/python/ref_string_count.asp

# EXAMPLE 1
txt = "I love apples, apple are my favorite fruit"
x = txt.count("apple")
print(x)

# EXAMPLE 2 : Search from position 10 to 20
color = "I used to hate pink, but now pink is my favorite color."
c = color.count("pink", 10, 20)
print(c)

# EXAMPLE 3 : Search from position 11 to 40
license = "To drive a car, a person must have a car license."
l = license.count("car", 11, 40)
print(l)
```

```python
# Python String endswith() Method
# The endswith() method returns True if the string ends with the specified value, otherwise False.

# https://www.w3schools.com/python/ref_string_endswith.asp

# EXAMPLE 1 : Check if the string ends with the phrase "my world.":
txt = "Hello, welcome to my world."
x = txt.endswith(".")
print(x)

# EXAMPLE 2 : Check if position 5 to 15 ends with the phrase "rapper.":
person = "Malique is a talented rapper."
p = person.endswith("rapper.", 5, 15)
print(p)

# EXAMPLE 3 : Check if the string ends with either the phrase "eat" or "more!"
more = "I want to eat some more!"
m = more.endswith(("eat","more!"))
print(m)
```

```python
# Python String find() Method
# The find() method finds the first occurrence of the specified value.
# The find() method returns -1 if the value is not found.

# https://www.w3schools.com/python/ref_string_find.asp

# EXAMPLE 1 : the text is the word "welcome"
txt = "Hello, welcome to my world."
x = txt.find("welcome")
print(x)

# EXAMPLE 2 : the first occurrence of the letter "a" when you only search between position 14 and 20
candy = "Kids loves to eat candy."
c = candy.find("a", 14, 20)
print(c)

# EXAMPLE 3 : If the value is not found, the find() method returns -1, but the index() method will raise an exception
mug = "That's my favorite mug!"
print(mug.find("z"))
print(mug.index("z"))
```

```python
# Python String islower() Method
# The islower() method returns True if all the characters are in lower case, otherwise False.

# https://www.w3schools.com/python/ref_string_islower.asp

# EXAMPLE 1
txt = "hello world!"
x = txt.islower()
print(x)

# EXAMPLE 2
meals = "my Father's tomyum is the best!"
m = meals.islower()
print(m)

# EXAMPLE 3
a = "ahmad loves coffee"
b = "Sara loves matcha"
c = "coffee and matcha contains caffien"

print(a.islower())
print(b.islower())
print(c.islower())
```

```python
# Python String isupper() Method
# The isupper() method returns True if all the characters are in upper case, otherwise False.

# https://www.w3schools.com/python/ref_string_isupper.asp

# EXAMPLE 1
txt = "THIS IS NOW!"
x = txt.isupper()
print(x)

# EXAMPLE 2
place = "DAMANSARA, PETALING Jaya"
p = place.isupper()
print(p)

# EXAMPLE 3
a = "HELLO, ANYBODY HOME?"
b = "My NAME IS Ahmad"

print(a.isupper())
print(b.isupper())
```

```python
# Python String lower() Method
# The lower() method returns a string where all characters are lower case.

# https://www.w3schools.com/python/ref_string_lower.asp

# EXAMPLE 1
txt = "Hello my FRIENDS"
x = txt.lower()
print(x)

# EXAMPLE 2
hike = "Lisa wants to hike Gunung Datuk."
h = hike.lower()
print(h)

# EXAMPLE 3
shoe = "Makerz is one of the Best LOCAL shoes brand."
s = shoe.lower()
print(s)
```

```python
# Python String upper() Method
# The upper() method returns a string where all characters are in upper case.

# https://www.w3schools.com/python/ref_string_upper.asp

# EXAMPLE 1
txt = "Hello my friends"
x = txt.upper()
print(x)

# EXAMPLE 2
tea = "Green tea is delicious."
t = tea.upper()
print(t)

# EXAMPLE 3
run = "I run 3KM everyday."
r = run.upper()
print(r)
```

```python
# Python String startswith() Method
# The startswith() method returns True if the string starts with the specified value, otherwise False.

# https://www.w3schools.com/python/ref_string_startswith.asp

# EXAMPLE 1
txt = "Hello, welcome to my world."
x = txt.startswith("Hello")
print(x)

# EXAMPLE 2 : Check if position 5 to 15 starts with the characters "Mal"
person = "Malique is a talented rapper."
p = person.startswith("Mal", 5, 15)
print(p)

# EXAMPLE 3 : Check if the string starts with either "I" or "You"
more = "I want to eat some more!"
m = more.startswith(("I","You"))
print(m)
```

```python
# Python String swapcase() Method
# The swapcase() method returns a string where all the upper case letters are lower case and vice versa.

# https://www.w3schools.com/python/ref_string_swapcase.asp

# EXAMPLE 1
txt = "Hello My Name Is PETER"
x = txt.swapcase()
print(x)

# EXAMPLE 2
place = "DAMANSARA, PETALING Jaya"
p = place.swapcase()
print(p)

# EXAMPLE 3
car = "The Toyota YARIS model attracts Many Youngsters."
c = car.swapcase()
print(c)
```

```python
# Python String replace() Method
# The replace() method replaces a specified phrase with another specified phrase.
# All occurrences of the specified phrase will be replaced, if nothing else is specified.

# https://www.w3schools.com/python/ref_string_replace.asp

# EXAMPLE 1
txt = "I like bananas"
x = txt.replace("bananas", "apples")
print(x)

# EXAMPLE 2 : Replace all occurrence of the word "pink"
color = "I used to hate pink, but now pink is my favorite color."
c = color.replace("pink", "blue")
print(c)

# EXAMPLE 3 : Replace the two first occurrence of the word "Datuk"
hike = "Lisa wants to hike Gunung Datuk. Gunung Datuk height 885 ftMSL and Gunung Datuk is too extreme for beginners."
h = hike.replace("Datuk", "Angsi", 2)
print(h)
```
