```python
# Functions In Python
# A function is a block of code which only runs when it is called.
# Functions provide better modularity for your application and a high degree of code reusing.
# Good thing about functions is that they are famous with several names. Different programming
# languages name them differently, for example, functions, methods, sub-routines, procedures, etc.
# Functions make your code cleaner and save you a lot of time.
# Python Functions stop us from writing the same logic various times.
# We can bind the logic in one function and then call the same repeatedly.

# In Python a function is defined using the def keyword

# Syntax
# def function_name():
#     code to be executed

# So, why should you learn to use functions?

# Here are 5 clear reasons:

# Avoid repetition (Don't Repeat Yourself – DRY)

# Make your code easier to read

# Easier to fix and debug

# Break complex problems into smaller parts

# Reuse code in different places or even in different programs
```

```python
while True:
  print('Enter Any Number or Type exit or quit to STOP')
  user_input = input()
  if user_input == 'exit' or user_input == 'quit':
    break
  else:
      user_input = int(user_input)
      if user_input%2 == 0:
        print('Even Number')
      else:
        print('Odd Number')
```

```python
# functions without parameter/arguments

# function definition
def greetUser():
  print('Welcome User')

# function calling
greetUser()
greetUser()
greetUser()
```

```python
# function with parameter/arguments

# what is a parameter
# a variable defined in a function definition is known as PARAMETER
# act as a placeholder

# function definition
def greetUser(name):                # name is a parameter
  print('Welcome User :',name)

# function calling
greetUser("John")                   # "John" is an argument   # arguments supplies data to a function
greetUser("Smith")                  # arguments are defined in function calling
```

```python
# def a function to check whether the entered no. is Even or Odd

def checkEvenOdd(n):
  if n % 2 == 0:
    print("Even")
  else:
    print("Odd")

num = int(input('Enter any number : '))
checkEvenOdd(num)
```

```python
# ANOTHER WAY USING RETURN
# function with parameter and with return statement
# a function with return statement is used to return a value

def checkEvenOdd(n):
  return n%2==0       # perform computation and return the "res"

num = int(input('Enter any number : '))
res = checkEvenOdd(num)

if res:
    print('Even')
else:
  print('Odd')
```

```python
# check whether the string is a palindrome or not
# madam madam
# nitin nitin

# Mr Yash's Solution
def strReverse(text):
  return text[::-1]

a = "madam"
res = strReverse(a)

if a == res:
  print("Palindrome String")
else:
  print("Not a Palindrome String")
```

```python
# check whether the string is a palindrome or not
# madam madam
# nitin nitin

def palindrome (p):
  return p[::] == p[::-1]

word = input ('Please enter any word : ')
res = palindrome (word)

if res:
    print ('Palindrome')
else:
    print ('Not Palindrome')
```

```python
# check whether the string is a palindrome or not
# madam madam
# nitin nitin

# Chatgpt's Solution: not using RETURN

def check_palindrome(words):

    for word in words:
        if word == word[::-1]:
            print(f"{word} → Palindrome ✅")
        else:
            print(f"{word} → Not Palindrome ❌")


mylist = ['madam', 'nitin', 'Python']

check_palindrome(mylist)
```
