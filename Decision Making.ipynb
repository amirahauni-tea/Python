```python
# Task
# Print The Largest
# multiple if read line by line & update the largest number

a = 19
b = 45
c = 27
d = 141
e = 67

largest = 0

if a>largest:     # 19>0 (t)
  largest = a     # largest = 19
if b>largest:     # 45>19 (t)
  largest = b     #largest = 45
if c>largest:     # 27>45 (f)
  largest = c     #largest = 45
if d>largest:     # 141>45 (t)
  largest = d     # largest = 141
if e>largest:     # 67>141 (f)
  largest = e     #largest = 141

print(f'Largest is : {largest}')
```

```python
# TASK 1 (DAY 4)
# Grading Allotment Applicaton
# if elif else reads the first line & ignore others if it's TRUE (SAVE TIME)

# Get input marks = ?

# Condition
# 100 - 85 : A Grade
# 84 - 75  : B Grade
# 74 - 60  : C Grade
# 59 - 40  : D Grade
# 39 - 0   : Fail

# -98 Invalid Marks
# 120 Invalid Marks

marks = int(input('Enter Any Number : '))
if marks <=100 and marks >=85:
  print('A Grade')
elif marks <=84 and marks >=75 :
  print('B Grade')
elif marks <=74 and marks >= 60:
  print('C Grade')
elif marks <=59 and marks >=40:
  print('D Grade')
elif marks <=40 and marks >=0:
  print('Fail')
else:
  print('Invalid Marks')
```

```python
# Chained Comparison
# helps us define a range, with only AND operator
# Lower limit <= value <= Upper Limit

print('Grading Alloment Application')
marks = int(input('Enter Your Marks : '))

# here 85 <= marks <= 100 means:
# a) the value is less than or equal to 100           c1    is 93 <= 100 (t)
# b) 85 means that I am less than the define value    c2    is 85 is less than 93 (t)

if 85 <= marks <= 100:
  print('Grade A')
elif 75 <= marks <= 84:
  print('Grade B')
elif 60 <= marks <= 74:
  print('Grade C')
elif 40 <= marks <= 59:
  print('Grade D')
elif 0 <= marks <= 39:
  print('Fail')
else:
  print('Invalid Marks')
```

```python
# TASK 2 (DAY 4)
# BMI CALCULATOR

# name = ?
# weight in kg = ?
# height in cm = ?

# convert cm in meters
# bmi = weight / meters*meters
#  Underweight < 18.5
#  Normal Weight 18.5 - 24.9
#  Overweight 25 - 29.9
#  Obese 30 and above

print("This is a BMI Calculator")

name = input("Enter your name : ")
weight_kg = float(input("Enter your weight in kg : "))
height_cm = float(input("Enter your height in cm : "))

height_m = height_cm / 100

bmi = weight_kg / (height_m*height_m)

if 0 <= bmi <= 18.5:
  print(f"Hi user {name} your BMI is {bmi}. You're Underweight 😕")
elif 18.5 <= bmi <= 24.9:
  print(f"Hi user {name} your BMI is {bmi}. You're Normal Weight 🤩")
elif 25 <= bmi <= 29.9:
  print(f"Hi user {name} your BMI is {bmi}. You're Overweight 😱")
else:
  print(f"Hi user {name} your BMI is {bmi}. You're Obese 😭")

```
