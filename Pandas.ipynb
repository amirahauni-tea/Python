```python
# Pandas Data Science Library

import numpy as np
import pandas as pd
print(pd.__version__)
```

```python
# numpy --> array()
arr = np.array([10,20,30,40,50])
print(type(arr))
print(arr)
```

```python
# Pandas Series
# Series is 1D Labelled Array

s = pd.Series([10,20,30,40,50])
print(type(s))
print(s)
print(s.shape)

# In Numpy --> Only Numbers
# In Series --> Numbers + Labels (index)
```

```python
info = pd.Series(['John',28,6700,'Klang'],index=['name','age','sal','city'])
print(info)
print(info['name'])         # indexing
print(info.get('city'))     # getting data values using get()
print(info.iloc[0])  # --> use .iloc instead of print(info[0])

# Series is like a Dictionary + array combo
```

```python
info = {'name':'John','age':28}
print(type(info))
print(info['name'])
# print(info[0])  --> dictionary didn't have index no.
```

```python
# values property
# print only values

print(info)   # --> series kind
print(info.values)  # --> numpy kind
```

```python
# Indexing Slicing and Using A For Loop

s = pd.Series([100,200,300,400,500,600],index=['v1','v2','v3','v4','v5','v6'])
print(s)

# 100   200   300   400   500   600
#  0     1     2     3     4     5
# -6    -5    -4    -3     -2   -1
# v1    v2    v3    v4    v5    v6

# Indexing in Series can be done by using [] or iloc[] or loc[]
print('\nIndexing in Series')
print(s.iloc[0])    # iloc used for index numbers
print(s.loc['v5'])  # loc used for labels
print(s['v2'])      # [] used for labels
print(s.iloc[-1])


print('\nSlicing in Series')
print(s[1:4])         # 4 will be excluded --> stopIndex
print(s['v1':'v5'])   # if performing slicing using labels, the end label is included

# 100 400 600
print(s[['v1','v4','v6']])  # for specific information [['','','']]
print(s[['v1','v4','v6']].values)

# conditional selection
print('\nConditional Selection')
print(s[s%3==0])    # divisible by 3
```

```python
#   Using for loop
for i in s:
  print(i)
```

```python
for k,v in s.items():
  print(k,v)
```

```python
for i in enumerate(s):
  print(i)
```
