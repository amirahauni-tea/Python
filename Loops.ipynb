```python
# Loops

# Loops in Python are used to repeat a block of code multiple times without
# writing the same code again and again.

# Loops help us repeat tasks efficiently and smartly.

# A loop is a control structure that repeats a block of code until a condition is met.

# for and while
# USE FOR LOOP WHEN THE NUMBER OF ITERATIONS ARE FIXED'
# OR USE FOR LOOP WHEN YOU'RE LOOKING FOR PERFORMING AN ITERATION

# USE WHILE LOOP WHEN THE NUMBER OF ITERATION ARE NOT FIXED
# WHEN THE NUMBER OF ITERATION ARE NOT KNOWN
```

```python
# for loop
# for loop is used to iterate over a Sequence [collection of data values]

# WHEN TO USE?
# when you want to perform an iteration and when you know how many times you are going to repeat your program, meaning when the number of iterations are known.

# syntax
# for iterable_variable in Sequence:
#    block of code

# AFTER EVERY ITERATTION BODY OF FOR LOOP
# WILL BE EXECUTED

for i in "Python":
  print(i)

# item 1 --> i = P
# item 2 --> i = y
# item 3 --> i = t
# item 4 --> i = h
# item 5 --> i = o
# item 6 --> i = n
```

```python
# NESTED FOR LOOP

# recommendation system
# which products people used to purchase together

# OUTER LOOP WILL WAIT UNTIL INNER FINISHES THE EXECUTION (c1,v1), (c1,v2), (c1,v3), (c2,v1), ....(c3,v3)

for c in ['red', 'blue', 'green']:    # OUTER FOR LOOP
  for v in ['car', 'bike', 'bus']:    # INNR FOR LOOP
    print(c,v)
```

```python
# while loop
# used to repeat a block of code
# until a certain condition is met

# WHEN TO USE?
# when the number of iterations are not known

# syntax
# initialization   [optional]
# while condition/expression:
# code to be executed
# increment / decrement [optional]

# WHILE CHECK THE CONDITION FIRST
# IF THE CONDITION IS TRUE, WHILE WILL EXECUTE ITS BODY
# AFTER WHILE FINISH EXECUTING THE BODY, CONDITION WILL BE CHECKED AGAIN
# IF THE CONDITION IS TRUE, WHILE WILL AGAIN EXECUTE ITS BODY

i = 1           # 1ST ITER i = 1               2nd ITER i = 2
while i <=3:    # 1<=3 (t) (Check condition)   2<=3 (t)
  print(i)      # 1 (Print)                    2
  i = i + 1     # i = 1+1 ==> 2 (Update)       i = 2+1 ==> 3
```

```python
# PRINT MATHEMATICAL TABLE OF A NUMBER 13
# USING WHILE LOOP

# 13X1 = 13
# 13X2 = 26

num = 13                            # 1st ITER: i = 1
i = 1                               # Check condition: 1<=10 (t)
while i <=10:                       # Print: (13X1 = 13)
  print(f"{num}X{i} = {num*i}")     # Update i: 1 + 1 = 2
  i = i + 1
```

```python
# DAY 6 (Video)

# palindrom of number   e.g. 1331 = 1331   66566 = 66566
# 1) divide the number by 10 and get its remainder
# 2) save the remainder by multiplying it with 10
# 3) update the number by dividing the number by 10

num = int(input('Enter a sequence of numbers  :'))
temp = num
sum = 0

while num != 0:         # 1st ITER  121!=0(t)               # 2nd ITER 12!=0(t)         # 3rd ITER 1!=0
  rem = num % 10        # rem = 1                           # rem = 2                   # rem = 1
  sum = sum*10 + rem    # sum = 0*10 + rem = 1              # sum = 1*10 + 2 = 12       # sum = 12*10 + 1 = 121
  num = num // 10       # num = 12                          # num = 1                   # num = 1//10 = 0
                        # (Float division operator //)

print(num,sum)
if temp == sum:       # use temp so Python will read a copy of num (temp=121)
  print('Palindrome Number')
else:
  print('Not A Palindrome')
```

```python
# break and continue statement

# break statement
# used to TERMINATE the execution of a LOOP

# break and continue should be defined inside decision making

for i in range(1,11):
  if i == 5:
    break
  print(i)
```

```python
# continue statement
# used to SKIP the current execution of LOOP

for i in range(1,11):
  if i == 5 or i == 9:      # skip 5 and 9
    continue
  print(i)
```

```python
# Guess a number game
# Objective: Our machines will generate one random number
# we have to guess the number in 4 attempts

import random
guessAttempt = 1

print('🎈 Guess A Number Game')
print('Enter Your Name')
name = input()
print(f'Welcome User {name} 😎')
print('I am thinking of a number between 1 and 13 🤔. Please Wait.....')
randomNumber = random.randint(1,13)
print('Number is Generated ✅. Guess The Number in 4 Attempts!')

while guessAttempt <= 4:
  print(f'{guessAttempt} Attempt. Take A Guess.')
  guess = int(input())

  if guess == randomNumber:
    print('🎉 Congratulations. Correct Guess 🎉')
    print(f'You have guess the number in {guessAttempt} attempt.')
    break

  if guess > randomNumber:
    print('Your Guess is Too HIGH ⬆️. Think of A Smaller Number')

  if guess < randomNumber:
    print('Your Guess is Too LOW ⬇️. Think of A Bigger Number')

  guessAttempt = guessAttempt + 1

if guess != randomNumber:
    print('❌GAME OVER! ❌')
    print(f'The Correct Number Was {randomNumber} ✳️')
```
