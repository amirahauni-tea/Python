```python
# Exception handling in Python is a mechanism that allows you to handle runtime errors,
# or exceptions, in a controlled way, rather than letting your program crash unexpectedly.
# Python provides a way to catch errors, respond to them, and allow your program to continue
# running smoothly even if something goes wrong.

# try except finally
# try : keep that block of code inside try , which might give you an error
# except : catch that error
```

```python
# Without Using Exception Handling

val1 = int(input('Enter First Value : '))
val2 = int(input('Enter Second Value  : '))
if val2>0:
  res = val1 / val2
  print(f'After {val1}/{val2} Division Result Will BE {res}')
else:
  print('Second Value Cannot Be Zero')

print('Rest of The Code')
```

```python
# Using Exception Handling

try:
  val1 = int(input('Enter First Value : '))
  val2 = int(input('Enter Second Value  : '))
  res = val1 / val2
  print(f'After Division Result Will Be {res}')
except:
  print('Something Goes Wrong....')
  print('Make Sure to Provide Only Integer Input or Second Value Can\'t Be Zero')
print('Rest Of The Code')
```

```python
# Using Exception Class
# if you need to know the exact reason for an error use Exception Class

try:
  val1 = int(input('Enter First Value : '))
  val2 = int(input('Enter Second Value  : '))
  res = val1 / val2
  print(f'After Division Result Will Be {res}')
except Exception as e:
  print('Something Goes Wrong....')
  print('Error :',e)
print('Rest Of The Code')
```

```python
# IndexError

list1 = [10,20,30,40]
#       0  1  2  3

try:
  print(list1[6])
except Exception as e:
  print('Error  :',e)
```

```python
# Using a Dedicated Exception Class

try:
  val1 = int(input('Enter First Value : '))
  val2 = int(input('Enter Second Value  : '))
  res = val1 / val2
  print(f'After Division Result Will Be {res}')
except ZeroDivisionError as e:
  print('Something Goes Wrong....')
  print('Error :',e)
print('Rest Of The Code')
```

```python
# Using a Dedicated Exception Class

try:
  val1 = int(input('Enter First Value : '))
  val2 = int(input('Enter Second Value  : '))
  res = val1 / val2
  print(f'After Division Result Will Be {res}')
except (ZeroDivisionError,ValueError) as e:
  print('Something Goes Wrong....')
  print('Error :',e)
print('Rest Of The Code')
```

```python
# finally
# finally execute always

# Case 1: WHEN THERE IS NO EXCEPTION
try:
  val1 = int(input('Enter First Value : '))
  val2 = int(input('Enter Second Value  : '))
  res = val1 / val2
  print(f'After Division Result Will Be {res}')
except Exception as e:
  print('Something Goes Wrong....')
  print('Error :',e)
finally:
  print('I WILL EXECUTE ALWAYS 😎')
print('Rest Of The Code')
```

```python
# Case 1: WHEN THERE IS AN EXCEPTION
try:
  val1 = int(input('Enter First Value : '))
  val2 = int(input('Enter Second Value  : '))
  res = val1 / val2
  print(f'After Division Result Will Be {res}')
except Exception as e:
  print('Something Goes Wrong....')
  print('Error :',e)
finally:
  print('I WILL EXECUTE ALWAYS 😎')
print('Rest Of The Code')
```

```python
# Case 1: WHEN THERE IS AN EXCEPTION BUT NOT HANDLED PROPERLY
try:
  val1 = int(input('Enter First Value : '))
  val2 = int(input('Enter Second Value  : '))
  res = val1 / val2
  print(f'After Division Result Will Be {res:2f}')
except ImportError as e:
  print('Something Goes Wrong....')
  print('Error :',e)
finally:
  print('I WILL EXECUTE ALWAYS 😎')
print('Rest Of The Code')
```
