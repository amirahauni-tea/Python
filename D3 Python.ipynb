```python
# TYPE CASTING
# A process of converting one data type to another data type

# automatic type conversion
# lower data type will be converted to higher data type automatically

# string(higher) --> float --> int(lower)

a = 10
b = 45.03
res = a + b
print(res)
```

```python
# manual type conversion EXPLICIT
# performed by developers

a = 10
b = 45.03
res = a + int(b)    # 10 + 45
print(res)

# floatVar --> intVar
# int(floatVar)   int(23.33) --> 23

# strVar --> intVar
# int(strVar)   int('789') --> 189

number = "789"
print(int(number))

# intVar --> floatVar
# float(intVar)   float(10) --> 10.0

# intVar --> strVar
# str(intVar)   str(24) --> '24'

# floatVar --> strVar
# str(floatVar)   str(24.334) --> '24.334'

# boolean --> int and float
choice = True
print(int(choice))
print(float(choice))

choice = False
print(int(choice))
print(float(choice))

# strVar --> floatVar
name = "89.21"
print(float(name))
```

```python
# INPUT FUNCTION AS STRING
# NO TYPE CASTING

val1 = input("Enter the First Value : ")    # '20'
val2 = input("Enter the Second Value : ")   # '27'
res = val1 + val2                           # res = 2027
print(f"After Addition {val1} + {val2} Result will be {res}")
```

```python
# INPUT function will get the value in the form of STRING
# perform TYPE CASTING (strVar --> intVar)

val1 = int(input("Enter the First Value : "))    # '20'
val2 = int(input("Enter the Second Value : "))   # '26'
res = val1 + val2                           # res = '20' + '26'
print(f"After Addition {val1} + {val2} Result will be {res}")
```

```python
 # BMI CALCULATOR

print("This is a BMI Calculator")

# Input from user
name = input("Enter your name : ")
weight_kg = float(input("Enter your weight in kg : "))
height_cm = float(input("Enter your height in cm : "))

# convert height in cm to m
height_m = height_cm / 100

# BMI = weight in kg / height in meters * height in meters
bmi = weight_kg / (height_m*height_m)
print(f"Hi user {name} your BMI is {bmi} 🤩")
```

```python
# Operators
# Used to perform operations on variables

# a) ARITHMATIC OPERATORS (+    -  *  /  //  **)
# Used to perform mathematical operations on variables

print(10+5)
print(10-5)
print(10*5)
print(10/5)    # In Python, DIVISION Operator used / to return value in FLOAT
print(10//5)   # In Python, FLOOR DIVISION Operator used // to return value in INT (ROUND OFF TO SMALL VALUE)
print(10%3)    # MODULUS used to give REMAINDER
print(2**8)    # 2 raise to the power of 8  ** Exponent
print(72**72)
```

```python
# b) ASSIGMENT OPERATOR
# used to assign a value to a variable

a = 90    # a as 90   a is 90
print(a)
```

```python
# c) COMPARISON OPERATORS
# <   >   ==    !=    <=    >=
# Used to return a BOOLEAN VALUE  [True / False]

print(10<5)
print(10>5)
print(98==98)
print(100!=1000)    # 100 not equal to 1000

age = 18
print(age >= 18)    # age should be equal or bigger than 18
```

```python
# d) LOGICAL OPERATORS
# works with multiple conditions
# and or not
# used ro return a BOOLEAN VALUE [TRUE / FALSE]

# and
# and requires more than one condition
# and will return TRUE when both or all the conditions are TRUE

print(100==100 and 200==200 and 300==300)
#       True          True        True

print(100==100 and 200==2400)
#       True       False
```

```python
# or
# or requires more than one condition
# or will return TRUE when both or all the conditions are TRUE
# or will return TRUE if atleast any condition is TRUE
# or will return FALSE when both or all the conditions are FALSE

print(100==100 or 200==200 or 300==300)
#       True          True        True

print(100==100 or 200==2400)
#       True       False

print(100==1000 or 200==2000)
#       False       False
```

```python
# not
# not to reverse the result

print(not(100==1000 and 200==2000))
```

```python
# e) Membership Operator
#    in   not in

# in
# used to return TRUE
# if the specified value exists in a Sequence

print("P" in "Python")
print("o" in "Python")
print("John" in ["John", "Alex", "Smith"])
print("x" in "Python")
print("p" in "Python")
```

```python
# not in
print("x" in "Python")
print("x" not in "Python")

# hey interpreter, i think x does not exist in Python
# yes, you're right

print("t" not in "Python")
# hey interpreter, i think t does not exist in Python
# no, you're wrong
```
