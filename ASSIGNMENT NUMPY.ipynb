```python
# Task 1: Salary Classification
# Replace salary with "High" if salary >= 60000
# Replace salary with "Medium" if salary >= 40000 and < 60000
# Otherwise "Low"

import numpy as np

salary = np.array([25000, 40000, 55000, 70000, 30000,90000])

result = np.where(
    salary >= 60000, "High",
    np.where(salary >= 40000, "Medium", "Low")
)
print(result)
```

```python
# Task 2: Product Discount
# If price >= 150 --> 20% Discount
# If price >= 50 and < 150  --> 10% Discount
# Otherwise no Discount

# Return final price

prices = np.array([120,80,40,200,60,30])

newPrices = np.where(
    prices > 150, prices*0.8,
    np.where(prices >=50, prices*0.9, prices)
)
print(newPrices)
```

```python
# Task 3: Temperature Category
# "Cold" if temp < 15
# "Warm" if 15 <= temp < 30
# "Hot" if temp >= 30

temp = np.array([10, 18, 25, 35, 5, 30])

res = np.where(
    temp < 15, "Cold",
    np.where(temp <30, "Warm", "Hot")
)
print(res)
```

```python
# Task 4: Even/Odd with Size Check
# Return "Big Even" if number >= 50 AND even
# Return "Odd" if Odd
# Otherwise "Small Even"

arr = np.array([10, 25, 60, 33, 80, 45])

result = np.where(
    (arr%2==0) & (arr>=50), "Big Even",
    np.where(arr%2!=0, "Odd", "Small Even")
)
print(result)
```

```python
# Task 5: Student Grades
# "A" if marks >= 85
# "B" if 70 <= marks < 85
# "C" if 50 <= marks < 70
# "Fail" if marks < 50

marks = np.array([35, 55, 72, 90, 48, 67])

grade = np.where(
    marks >= 85, "A",
    np.where(marks >= 70, "B",
             np.where(marks >= 50, "C", "Fail"))
)
print(grade)
```
